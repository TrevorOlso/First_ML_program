```python
##Setting up data


#import necessary tools
import numpy as np 
import pandas as pd 
import os
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score


#read files into program
gender_submission = pd.read_csv("/kaggle/input/titanic/gender_submission.csv")
test = pd.read_csv("/kaggle/input/titanic/test.csv")
train = pd.read_csv("/kaggle/input/titanic/train.csv")

#fill in missing data points
avg_age = round(train['Age'].mean())
train = train.fillna({"Age": avg_age})
train = train.fillna({"Embarked": "S"})
test = test.fillna({"Age": avg_age})
test = test.fillna({"Embarked": "S"})



#getting rid of unnecesary info
train = train.drop(['Cabin'] , axis = 1)
train = train.drop(['Ticket'] , axis = 1)
train = train.drop(['Name'] , axis = 1)
train = train.drop(['Fare'] , axis = 1)

test = test.drop(['Cabin'] , axis = 1)
test = test.drop(['Ticket'] , axis = 1)
test = test.drop(['Name'] , axis = 1)
test = test.drop(['Fare'] , axis = 1)


#changing male and female to bianary and for training model
map_sex =  {"male": 1 , "female": 0}
train['Sex'] = train['Sex'].map(map_sex)
test['Sex'] = test['Sex'].map(map_sex)

map_embarked = {"S": 1 , "C": 2 , "Q": 3}
train['Embarked'] = train['Embarked'].map(map_embarked)
test['Embarked'] = test['Embarked'].map(map_embarked)


test.info()
```

```python
#setting values for training
target = train['Survived']
predict = train.drop(['Survived' , 'PassengerId'] , axis = 1)


#training
x_train, x_val, y_train, y_val = train_test_split(predict, target, test_size=0.4, random_state=0)
rf = RandomForestClassifier()
rf.fit(x_train, y_train)
y_pred = rf.predict(x_val)

#calculating accuracy of model
accuracy = accuracy_score(y_pred , y_val)
print(round(accuracy * 100, 1))

#outputing data
predictions = rf.predict(test.drop(['PassengerId'] , axis = 1))
output = pd.DataFrame({'PassengerId': test['PassengerId'], 'Survived': predictions})
output.to_csv("titanic_predictions.csv" , header = 'predictions' , index = False)
```
